```python
class Tree:
    def __init__(self, data):
        self.data = data
        self.children = []
        self.parent = None

    def add_child(self, child):
        child.parent = self
        self.children.append(child)

    def __str__(self):
        return f"Hey i am a Tree"

    def get_level(self):
        level = 0
        p = self.parent
        while p is not None:
            level += 1
            p = p.parent
        return level

    def print_tree(self):
        spc = self.get_level() * 3
        prefix = "|" + ("_" * spc)
        print(f"{prefix}{self.data}")
        if self.children:
            for i in self.children:
                i.print_tree()
```

```python
root = Tree("Electronics")

laptop = Tree("Laptop")
asus = Tree("Asus")

laptop.add_child(asus)
# leaf node
laptop.add_child(Tree("MAC"))
# leaf node
laptop.add_child(Tree("Acer"))

asus.add_child(Tree("Asus Rog 1"))
asus.add_child(Tree("Asus Rog 2"))

smartphone = Tree("Smartphone")
iphone = Tree("iphone")

smartphone.add_child(iphone)
# leaf node
smartphone.add_child(Tree("Xiaomi"))
# leaf node
smartphone.add_child(Tree("Samsung"))

iphone.add_child(Tree("Iphone 11"))
iphone.add_child(Tree("Iphone 12"))
iphone.add_child(Tree("Iphone 13"))
iphone.add_child(Tree("Iphone 14"))

root.add_child(laptop)
root.add_child(smartphone)
```

```python
def trial(x):
    if x == 1:
        return
    print(x * (x - 1))
    x = x - 1
    trial(x)
```

```python
# x -> x^x

# Tail Recursion

def Factorial(x, ct=0):
    if ct == x:
        return 1
    return x * Factorial(x, ct=ct + 1)
```

```python
Factorial(4)
```

```python
# Non Tail Recursion


def Factorial(x):
    if x == 1:
        return 12
    print(x * Factorial(x - 1))
    return 13
```

```python
Factorial(4)
```

```python
# Recursion in a function

def GetDetails(name, age, mobile, ct=0):
    if ct == 5:
        return "Maximum Reached"
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Mobile: {mobile}")
    name = input()
    age = input()
    mobile = input()
    GetDetails(name, age, mobile, ct=ct + 1)
```

```python
# Iterative
for i in range(5):
    name = input()
    age = input()
    mobile = input()
    GetDetails(name, age, mobile)
```

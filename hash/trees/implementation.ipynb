```python
class Tree:
    def __init__(self, data):
        self.data = data
        self.children = []
        self.parent = None

    def add_child(self, child):
        child.parent = self
        self.children.append(child)

    def __str__(self):
        return f"Hey i am a Tree"
```

```python
root = Tree("Electronics")

laptop = Tree("Laptop")
laptop.add_child(Tree("Asus"))
laptop.add_child(Tree("MAC"))
laptop.add_child(Tree("Acer"))

smartphone = Tree("Smarthphone")
smartphone.add_child(Tree("iphone"))
smartphone.add_child(Tree("Xiami"))
smartphone.add_child(Tree("Samsung"))

root.add_child(laptop)
root.add_child(smartphone)
```

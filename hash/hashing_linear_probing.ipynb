```python
# key = []
# value = []

# 'pyhon' -> Key:45 -> Value


class Dictionary:
    def __init__(self, size):
        self.size = size
        # slots is the key array
        self.slots = [None] * size
        # data is the value array
        self.data = [None] * size

    def put(self, key, value):
        idx = self.hash_fx(key)
        if self.slots[idx] == None:
            self.slots[idx] = key
            self.data[idx] = value
        else:
            if self.slots[idx] == key:
                self.slots[idx] = key
                self.data[idx] = value
            else:
                if None not in self.slots:
                    return "Full"
                new_hash = self.rehash(idx)
                print(f"1st hash {new_hash}")
                ct = 1
                while self.slots[new_hash] is not None:
                    new_hash = self.rehash(new_hash)
                    ct += 1
                    print(f"{ct} hash {new_hash}")
                self.slots[new_hash] = key
                self.data[new_hash] = value

    def hash_fx(self, key):
        return abs(hash(key)) % self.size

    def rehash(self, old_hash):
        return (old_hash + 1) % self.size
```

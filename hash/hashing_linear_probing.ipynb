```python
import logging

# Configure logging
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="dictionary.log",  # Save logs to a file
    filemode="w",  # Overwrite log file on each run
)


class Dictionary:
    def __init__(self, size):
        self.size = size
        self.slots = [None] * size  # Key array
        self.data = [None] * size  # Value array
        self.count = 0  # Number of key-value pairs stored
        logging.info(f"Initialized Dictionary with size {self.size}")

    def put(self, key, value):
        """Insert a key-value pair into the dictionary."""
        logging.debug(f"Attempting to put key: {key}, value: {value}")
        if self.count == self.size:  # Check if the dictionary is full
            logging.warning("Dictionary full. Resizing...")
            self.resize()  # Resize the dictionary before adding

        idx = self.hash_fx(key)  # Compute the initial hash
        logging.debug(f"Initial hash index for key '{key}': {idx}")

        while self.slots[idx] is not None:
            if self.slots[idx] == key:  # Update value if key already exists
                self.data[idx] = value
                logging.info(
                    f"Updated key '{key}' with new value '{value}' at index {idx}"
                )
                return
            idx = self.rehash(idx)  # Find the next slot
            logging.debug(f"Collision occurred. Probing to next index: {idx}")

        # Insert the key-value pair
        self.slots[idx] = key
        self.data[idx] = value
        self.count += 1  # Increment the count of stored items
        logging.info(f"Inserted key '{key}' with value '{value}' at index {idx}")

    def get(self, key):
        """Retrieve the value associated with a key."""
        logging.debug(f"Attempting to get value for key: {key}")
        idx = self.hash_fx(key)
        start_idx = idx  # Save the starting index to detect a full cycle

        while self.slots[idx] is not None:
            if self.slots[idx] == key:  # If the key matches, return the value
                logging.info(
                    f"Key '{key}' found at index {idx} with value '{self.data[idx]}'"
                )
                return self.data[idx]
            idx = self.rehash(idx)  # Check the next slot
            logging.debug(f"Probing to next index: {idx}")
            if idx == start_idx:  # If we loop back to the start, key is not present
                break

        logging.warning(f"Key '{key}' not found")
        return "Not Found"  # Return None if the key is not found

    def hash_fx(self, key):
        """Hash function to compute the index."""
        index = abs(hash(key)) % self.size
        logging.debug(f"Computed hash index for key '{key}': {index}")
        return index

    def rehash(self, old_hash):
        """Rehash function for linear probing."""
        new_hash = (old_hash + 1) % self.size
        logging.debug(f"Rehashed index: {old_hash} -> {new_hash}")
        return new_hash

    def resize(self):
        """Double the size of the dictionary and rehash all existing key-value pairs."""
        logging.info("Resizing dictionary...")
        old_slots = self.slots
        old_data = self.data

        self.size *= 2  # Double the size
        self.slots = [None] * self.size
        self.data = [None] * self.size
        self.count = 0  # Reset count since we will re-add elements

        for i in range(len(old_slots)):
            if old_slots[i] is not None:  # Rehash existing key-value pairs
                logging.debug(
                    f"Rehashing key '{old_slots[i]}' with value '{old_data[i]}'"
                )
                self.put(old_slots[i], old_data[i])

        logging.info(f"Resizing complete. New size: {self.size}")

    def display(self):
        """Display the current state of the dictionary."""
        logging.info("Displaying current state of the dictionary")
        print("Slots:", self.slots)
        print("Data:", self.data)
        print("Size:", self.size)
        print("Count:", self.count)
```

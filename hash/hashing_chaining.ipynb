```python
class EmptyDtype(Exception):
    pass


class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.next = None

    def __str__(self):
        return f"{Node.__name__}"


class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return self.n

    def __str__(self):
        return f"{LinkedList.__name__}"

    def append(self, key, value):
        if self.head is None:
            new_node = Node(key, value)
            self.head = new_node
            self.n += 1
            self.__show()
        else:
            new_node = Node(key, value)
            curr = self.head
            while curr.next is not None:
                curr = curr.next
            curr.next = new_node
            self.n += 1
            self.__show()

    def __show(self):
        if self.head == None:
            raise EmptyDtype("List is Empty,Consider Prepending some data.")
        curr = self.head
        while curr.next != None:
            print(f"{curr.key}:,{curr.value}", end=" ")
            curr = curr.next
        print(f"{curr.key}:,{curr.value}=> None")

    def find_element(self, key):
        curr = self.head
        while curr is not None:
            if curr.key == key:
                return f"Element is {curr.value}"
            curr = curr.next
        return "Element is not in the list"

    # def drop(self):
    #     self.__init__()
    #     return f"{LinkedList.__name__} is Empty now"

    # def remove(self, data):
    #     curr = self.head
    #     if curr is None:
    #         raise Exception(
    #             "List is already empty. Consider adding some values using prepend or append."
    #         )

    #     # Handle the case where the value to be removed is at the head
    #     if curr.value == data:
    #         self.head = curr.next
    #         self.n -= 1
    #         self.__show()
    #         return

    #     # Traverse the list to find and remove the node
    #     while curr.next is not None:
    #         if curr.next.value == data:
    #             # Remove the node by updating the pointer
    #             curr.next = curr.next.next
    #             self.n -= 1
    #             self.__show()
    #             return
    #         curr = curr.next

    #     # Value not found in the list
    #     return "Value is not in the list !!"


class Dictionary:
    def __init__(self, size):
        self.size = size
        self.buckets = [LinkedList() for _ in range(self.size)]
```

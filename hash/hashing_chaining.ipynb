```python
class EmptyDtype(Exception):
    pass


class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.next = None

    def __str__(self):
        return f"{Node.__name__}"


class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return self.n

    def __str__(self):
        return f"{LinkedList.__name__}"

    def append(self, key, value):
        if self.head is None:
            new_node = Node(key, value)
            self.head = new_node
            self.n += 1
            self.__show()
        else:
            new_node = Node(key, value)
            curr = self.head
            while curr.next is not None:
                curr = curr.next
            curr.next = new_node
            self.n += 1
            self.__show()

    def __show(self):
        if self.head == None:
            raise EmptyDtype("List is Empty,Consider Prepending some data.")
        curr = self.head
        while curr.next != None:
            print(f"{curr.key}:,{curr.value}", end=" ")
            curr = curr.next
        print(f"{curr.key}:,{curr.value}=> None")

    def find_element(self, key):
        curr = self.head
        while curr is not None:
            if curr.key == key:
                return f"Element is {curr.value}"
            curr = curr.next
        return "Element is not in the list"


class Dictionary:
    def __init__(self, size=4):
        self.size = size
        self.buckets = [LinkedList() for _ in range(self.size)]
        self.total_elements = 0  # Track the total number of elements in the dictionary

    def hash_fx(self, key):
        index = abs(hash(key)) % self.size
        return index

    def rehash(self):
        print("Load Factor Crossed, Implementing Rehashing...")
        old_buckets = self.buckets
        self.size *= 2  # Double the size
        self.buckets = [LinkedList() for _ in range(self.size)]
        self.total_elements = 0  # Reset total elements and reinsert

        for bucket in old_buckets:
            curr = bucket.head
            while curr is not None:
                self.put(curr.key, curr.value, rehashing=True)
                curr = curr.next

    def put(self, key, value, rehashing=False):
        idx = self.hash_fx(key)  # Compute the hash
        curr = self.buckets[idx].head

        # Check if the key is already present in the bucket
        while curr is not None:
            if curr.key == key:
                curr.value = value
                if not rehashing:
                    print(f"Updated key '{key}' with new value '{value}'.")
                return
            curr = curr.next

        # Key is not found; add a new node
        self.buckets[idx].append(key, value)
        if not rehashing:
            self.total_elements += 1  # Increment only if not rehashing
            print(f"Added key '{key}' with value '{value}'.")

        # Check load factor and rehash if necessary
        load_factor = self.total_elements / self.size
        if load_factor > 0.75 and not rehashing:
            self.rehash()

    def search(self, key):
        idx = self.hash_fx(key)  # Compute the hash
        curr = self.buckets[idx].head
        iterations = 0  # Initialize counter for iterations

        while curr is not None:
            iterations += 1
            if curr.key == key:
                print(
                    f"Found key '{key}' with value '{curr.value}' after {iterations} iterations."
                )
                return curr.value
            curr = curr.next

        print(f"Key '{key}' not found after {iterations} iterations.")
        return None  # Return None if the key is not found
```

```python
class Dictionary:
    def __init__(self, size=4, load_factor_threshold=0.75):
        self.size = size
        self.buckets = [LinkedList() for _ in range(self.size)]
        self.total_elements = 0
        self.load_factor_threshold = load_factor_threshold

    def hash_fx(self, key):
        return (sum(ord(c) for c in str(key)) * 31) % self.size

    def next_prime(self, n):
        def is_prime(k):
            if k < 2:
                return False
            for i in range(2, int(k**0.5) + 1):
                if k % i == 0:
                    return False
            return True

        while not is_prime(n):
            n += 1
        return n

    def rehash(self):
        print("Load Factor Crossed, Implementing Rehashing...")
        old_buckets = self.buckets
        self.size = self.next_prime(self.size * 2)  # Increase to nearest prime
        self.buckets = [LinkedList() for _ in range(self.size)]
        self.total_elements = 0

        for bucket in old_buckets:
            curr = bucket.head
            while curr is not None:
                self.put(curr.key, curr.value, rehashing=True)
                curr = curr.next

    def put(self, key, value, rehashing=False):
        idx = self.hash_fx(key)
        curr = self.buckets[idx].head

        while curr is not None:
            if curr.key == key:
                curr.value = value
                if not rehashing:
                    print(f"Updated key '{key}' with new value '{value}'.")
                return
            curr = curr.next

        self.buckets[idx].append(key, value)
        if not rehashing:
            self.total_elements += 1
            print(f"Added key '{key}' with value '{value}'.")

        load_factor = self.total_elements / self.size
        if load_factor > self.load_factor_threshold and not rehashing:
            self.rehash()

    def search(self, key):
        idx = self.hash_fx(key)
        curr = self.buckets[idx].head
        iterations = 0

        while curr is not None:
            iterations += 1
            if curr.key == key:
                print(
                    f"Found key '{key}' with value '{curr.value}' after {iterations} iterations."
                )
                return curr.value
            curr = curr.next

        print(f"Key '{key}' not found after {iterations} iterations.")
        return None

    # Add this method to enable iteration
    def __iter__(self):
        for bucket in self.buckets:
            curr = bucket.head
            while curr is not None:
                yield curr.key, curr.value
                curr = curr.next

    # Enable indexing with square brackets
    def __getitem__(self, key):
        value = self.search(key)
        if value is None:
            raise KeyError(f"Key '{key}' not found in dictionary.")
        return value

    def __setitem__(self, key, value):
        self.put(key, value)

    # Enable key existence check with `in`
    def __contains__(self, key):
        return self.search(key) is not None
```
